```python
# Celcius to Farenheit Code
```

```python
temp_celsius = float(input("Enter temperature in Celsius: "))
temp_fahrenheit = (temp_celsius * 9/5) + 32
print(f"{temp_celsius} degrees Celsius is equal to {temp_fahrenheit} degrees Fahrenheit.")
if temp_fahrenheit <= 73:
  print("The temperature is sub zero")
elif temp_fahrenheit <= 103:
  print("The temperature is average")
else:
  print("Temperature is high")
```
